Problem 2: The Water Jug Problem (Solved using BFS)
Problem Statement:
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque

CAPACITY_A = 4
CAPACITY_B = 3
TARGET = 2


def get_next_states(state):
    a, b = state
    states = []

    # 1. Fill Jug A
    states.append(((CAPACITY_A, b), "Fill Jug A"))

    # 2. Fill Jug B
    states.append(((a, CAPACITY_B), "Fill Jug B"))

    # 3. Empty Jug A
    states.append(((0, b), "Empty Jug A"))

    # 4. Empty Jug B
    states.append(((a, 0), "Empty Jug B"))

    # 5. Pour A -> B
    amount = min(a, CAPACITY_B - b)
    states.append(
        ((a - amount, b + amount), "Pour Jug A -> Jug B")
    )

    # 6. Pour B -> A
    amount = min(b, CAPACITY_A - a)
    states.append(
        ((a + amount, b - amount), "Pour Jug B -> Jug A")
    )

    return states


def water_jug_bfs():
    start = (0, 0)

    queue = deque([(start, [])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        a, b = state

        # Check target
        if a == TARGET or b == TARGET:
            return path + [("Target Reached", state)]

        for next_state, operation in get_next_states(state):

            if next_state not in visited:
                visited.add(next_state)

                new_path = path + [
                    (operation, next_state)
                ]

                queue.append((next_state, new_path))

    return None


solution = water_jug_bfs()

print("Minimum sequence of steps:\n")

for i, (operation, state) in enumerate(solution):
    print(f"Step {i}: {operation}")
    print(f"       State = {state}")
```
